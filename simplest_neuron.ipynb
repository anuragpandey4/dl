```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42)
input = torch.tensor([3.0,2.8,2.7])
```

```python
neuron = nn.Linear(in_features=3,out_features=1)
```

```python
output = neuron(input)
```
